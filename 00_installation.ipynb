`````python
data_set_type = 'full'
args = Params()
args.root  = f'{DATAROOT}/Imagenet_{data_set_type}' # Directory containing images
`````

`````python
import os
from PIL import Image

# data_dir = "/Volumes/SSD1TO/DeepLearningDatasets/Imagenet_full/train"
for folder in args.folders :
    # first level
    data_dir = os.path.join(args.root, folder)
    print(f'\nFolder \"{folder}\" in path \"{data_dir}\":')
    for root, _, files in os.walk(data_dir):
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                path = os.path.join(root, file)
                try:
                    Image.open(path).verify()
                except Exception as e:
                    print(f"Invalid image: {path} ({e})")
`````

Note that we will use the localisation data from https://www.kaggle.com/c/imagenet-object-localization-challenge/data 


5.D Animal 10k  Data set

For the Animal 10k data set, described in the original [paper](https://arxiv.org/abs/2108.12617) follow the instructions on the corresponding [git](https://github.com/AlexTheBad/AP-10K/tree/main?tab=readme-ov-file#download)

Then gather the ``ap-10k.zip`` files into the folder defined as the `DATAROOT` parameter in the `retinotopy.py` file and extract the archive 

The path is slightly different using this data set, a `animal_10k` folder located in the `DATAROOT` hold both `annotations` for the data set and a `ap-10k/animal_data` for the images 

Note that in this case a single test data set is available (in opposition with the `test`, `train`, `val` Imagenet folders)
