```python
# The dataset to import images from

args = Params()

df_filename = f"{data_cache}/get_distance_to_plot_imagenet.parquet"

if not os.path.isfile(df_filename):
    print(f'{df_filename=}')
    df_dist = None

    annotations = get_annotation('csv')
    for image_name in tqdm(annotations['ImageId']):                  
        
        ground_true_indices, three_points, ground_true, origin_size = get_ground_true(args, image_name, annotations, 'Imagenet')
        
        if ground_true.min() == 0.0:
            dist = [euclidean_distance(three_points[0], three_points[1]), euclidean_distance(three_points[1], three_points[2])]
            dist[1] += dist[0]
            df_dist_ = pd.DataFrame({'ImageId':image_name, 'Distance_0': [dist[0]], 'Distance_1': [dist[1]]})
            df_dist = store_pandas(df_dist, df_dist_)
   
    df_dist.to_parquet(df_filename)
    print('All done sir !')
```
